### Tabla de Frecuencia

Una forma de describir una variable es con una tabla de frecuencia, que contiene una variable y su frecuencia (la cantidad de veces que ese valor aparece). Esta descripcion se conoce como la distribucion de una variable

```python
from os.path import basename, exists


def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + local)


download("https://github.com/AllenDowney/ThinkStats/raw/v3/nb/thinkstats.py")

from empiricaldist import FreqTab
from thinkstats import decorate
```

```python
t = [1.0, 2.0, 2.0, 3.0, 5.0]
```

```python
ftab = FreqTab.from_seq(t)
ftab
```

```python
# FreqTab proporciona un metodo llamado bar que grafica la frecuencia como un grafico de barras
ftab.bar()
decorate(xlabel = "Value", ylabel = "Frequency")
```

```python
# Al ser una seria de pandas poodemos indexarlo por columna
ftab[2.0]
```

```python
# O de esta manera
ftab(2.0)
```

```python
# Al ser una tabla de frecuencias cualquier valor que no aparece en la secuencia original tiene una frecuencia de 0
ftab(4.0)
```

```python
# FreqTab tiene un atributo llamado fs que devuelve un arra de frecuencias
ftab.fs
```

```python
for x, freq in ftab.items():
    print(x, freq)
```

### Outliers

Mirando tablas de frecuencias es facil identificar la forma de una distribucion y sus cantidades mas comunes, pero los valores atipicos no se ven facilmente. Antes de realizar algo, es una buena idea identificar outliers, que son valores extremos que pueden ser medidas o errores dentro de proceso, o hasta valores legitimos de eventos ocasionales.

```python
# Para identificar otuliers, sliceamos FreqTab para seleccionar los "n" valores mas pequeños y sus frecuencias
```

```python
def smallest(fatb, n = 10):
    return ftab.iloc[:n]
```

```python
smallest(ftab_length)
```

```python
def largest(ftab, n=10):
    return ftab.iloc[-n:]
```

```python
largest(ftab_length)
```

Algunos de estos valores probablemente sean errores. La mayoria de los medicos recomienda inducir el parto si el embarazo supera las 41 semanas, asi que 50 semanas parece poco probable. Pero no hay una linea clara entre los valores que son errores seguros y los que podrian ser casos raros pero correctos.

La mejor forma de manejar los valores atipicos depende del “conocimiento del dominio”, es decir, informacion sobre de donde vienen los datos y que significan. Tambien depende del analisis que se planee realizar.

En este ejemplo, la pregunta principal es si los primeros bebes tienden a nacer antes o despues que los demas. Por eso usaremos estadisticas que no se vean muy afectadas por un pequeño numero de valores incorrectos.

### Tamaño de efecto

Una diferencia de este tipo a veces es conocida como "efecto". Existen varias maneras de cuantificar la magnitud del efecto. La mas simple es reportar la diferente en terminos absolutos, o sea, en nuestro caso 0.078 semanas.

Otra forma es en terminos relativos. Por lo que podemos decir que primeros embarazos son 0.2% mas largos que otros, en promedio

```python
diff / live["prglngth"].mean() * 100
```

Otra opcion es informar un tamaño de efecto estandarizado, que es una estadistica pensada para cuantificar la magnitud de un efecto de forma que sea comparable entre diferentes cantidades y grupos.

Estandarizar significa expresar la diferencia como un multiplo de la desviacion estandar.

```python
diff / live["prglngth"].std()
```

Pero nota que usamos ambos grupos para calcular la desviacion estandar. Si los grupos son muy diferentes, la desviacion estandar combinada es mayor que la de cualquiera de los grupos por separado, lo que podria hacer que el tamaño del efecto parezca pequeño.

Una alternativa es usar la desviacion estandar de un solo grupo, pero no esta claro de cual. Podriamos tomar el promedio de las dos desviaciones estandar, pero si los grupos tienen tamaños diferentes, eso daria demasiado peso a uno y poco al otro.

Una solucion comun es usar la desviacion estandar combinada (pooled standard deviation), que es la raiz cuadrada de la varianza combinada (pooled variance), la cual es la suma ponderada de las varianzas de los grupos. Para calcularla, empezamos con las varianzas:

```python
import numpy as np
```

```python
group1, group2 = firsts["prglngth"], others["prglngth"]
```

```python
v1, v2 = group1.var(), group2.var()
```

```python
n1, n2 = group1.count(), group2.count()
pooled_var = (n1 * v1 + n2 * v2) / (n1 + n2)
```

```python
np.sqrt(pooled_var)
```

```python
# La desviacion estandar combinada esta entre la desviacion estandar de los grupos
firsts["prglngth"].std(), others["prglngth"].std()
```

Un tamaño de efecto estandarizaco usa la desviacion estandar combinada es llamado Cohen's effect size. Esta es una funcion que lo computa

```python
def cohen_effect_size(group1, group2):
    diff = group1.mean() - group2.mean()

    v1, v2 = group1.var(), group2.var()
    n1, n2 = group1.count(), group2.count()
    pooled_var = (n1 * v1 + n2 * v2) / (n1 + n2)

    return diff / np.sqrt(pooled_var)
```

### Glosario

#### distribución  
El conjunto de valores y la frecuencia con que cada valor aparece en un conjunto de datos.

#### tabla de frecuencias  
Una correspondencia entre valores y frecuencias.

#### frecuencia  
El número de veces que un valor aparece en una muestra.

#### sesgada  
Una distribución está sesgada si es asimétrica, con valores extremos que se extienden más en una dirección que en la otra.

#### moda  
La cantidad más frecuente en una muestra, o una de las más frecuentes.

#### distribución uniforme  
Una distribución en la que todas las cantidades tienen la misma frecuencia.

#### valor atípico  
Un valor extremo en una distribución.

#### estandarizado  
Una estadística está estandarizada si se expresa en términos que son comparables entre diferentes conjuntos de datos y dominios.

#### desviación estándar combinada  
Una estadística que combina datos de dos o más grupos para calcular una desviación estándar común.

#### tamaño del efecto de Cohen  
Una estadística estandarizada que cuantifica la diferencia en las medias de dos grupos.

#### significancia práctica  
Un efecto tiene significancia práctica si es lo suficientemente grande como para importar en la práctica.
